```python
import pandas as pd
import pickle
import numpy as np
```

```python
df = pickle.load(open("dataset.pkl",'rb'))
```

```python
df
```

```python
df.isna().sum()
```

```python
df['venue'].apply(lambda x: x.split()[0])
```

```python
df[df['city'].isna()]['venue'].value_counts()
```

```python
cities = np.where(df['city'].isna(),df['venue'].str.split().apply(lambda x:x[0]),df['city'])
```

```python
df['city'] = cities
```

```python
df.isna().sum()
```

```python
df.drop(columns=['venue'],inplace=True)
```

```python
df
```

```python
eligible_cities = df['city'].value_counts()[df['city'].value_counts() > 600].index.to_list()
```

```python
eligible_cities
```

```python
df = df[df['city'].isin(eligible_cities)]
```

```python
df
```

```python
df['score'] = df.groupby('match_id')['runs'].cumsum()
```

```python
df
```

```python
df['Over'] = df['ball'].apply(lambda x: str(x).split(".")[0])
df['ball'] = df['ball'].apply(lambda x: str(x).split(".")[1])
```

```python
df
```

```python
df['total_ball_bowled'] = df['Over'].astype(int)*6 + df['ball'].astype(int)
```

```python
df
```

```python
df['balls_left'] = 120 - df['total_ball_bowled']

df['balls_left'] =  df['balls_left'].apply(lambda x: abs(x) if x < 0 else x)
```

```python
df
```

```python
df['player_dismissed'] = df['player_dismissed'].apply(lambda x:0 if x=='0' else 1)
```

```python
df['player_dismissed'] = df['player_dismissed'].astype('int')
```

```python
df['player_dismissed'].info()
```

```python
df['player_dismissed'] = df.groupby('match_id')['player_dismissed'].cumsum()
```

```python
df['wickets_left'] = 10 - df['player_dismissed']
```

```python
df
```

```python
df['crr'] = (df['score']*6)/df['total_ball_bowled']
```

```python
df
```

```python
df.groupby('match_id')['runs'].rolling(window = 30).sum().values
```

```python
df['last_five'] = df.groupby('match_id')['runs'].rolling(window = 30).sum().values
```

```python
df
```

```python
final_df = df.groupby('match_id').sum()['runs'].reset_index().merge(df,on='match_id')
```

```python
final_df=final_df[['batting_team','bowling_team','city','score','balls_left','wickets_left','crr','last_five','runs_x']]
```

```python
final_df.isna().sum()
```

```python
final_df.dropna(inplace = True)
```

```python
final_df
```

```python
# reshuffling to prevent develop any kind of bias
 
final_df = final_df.sample(final_df.shape[0])
```

```python
final_df.sample(5)
```

```python
X = final_df.drop('runs_x', axis =1 )
y = final_df['runs_x']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.metrics import r2_score, mean_absolute_error
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from xgboost import XGBRegressor
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.25, random_state= 42)
```

```python
X_train
```

```python
transformer = ColumnTransformer(
    [
        ("ohe", OneHotEncoder(sparse_output= False, drop= "first", handle_unknown= "ignore"), ['batting_team', 'bowling_team', 'city']),
        ("sc", StandardScaler(),['score', 'balls_left', 'wickets_left', 'crr', 'last_five'])
    ], remainder= 'passthrough'
)
```

```python
model = XGBRegressor(n_estimators= 1000,learning_rate= 0.2, max_depth= 12,random_state= 42)
```

```python
pipe = Pipeline(steps= [
    ('step1', transformer),
    ('step2', model )
])
```

```python
pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
print(r2_score(y_test,y_pred))
print(mean_absolute_error(y_test,y_pred))
```

```python
pickle.dump(pipe,open('model.pkl', 'wb'))
```
